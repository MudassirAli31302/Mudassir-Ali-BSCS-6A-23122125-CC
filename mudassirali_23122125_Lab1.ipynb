```python
print("--- UNIVERSITY ADMISSION ELIGIBILITY ---")

courses = {
    "cs": (70, "CS", "Computer Science"),
    "engineering": (80, "Engineering", "Engineering"),
    "medical": (90, "Medical", "Medical"),
}

course = input("Enter course (CS / Engineering / Medical): ").strip().lower()
try:
    marks = int(input("Enter your entry test marks: "))
except ValueError:
    exit("Invalid marks. Please enter a number.")
if not 0 <= marks <= 100:
    exit("Invalid marks. Please enter marks between 0 and 100.")

if course not in courses:
    exit("Invalid option. Please choose CS, Engineering, or Medical.")

cutoff, short, full = courses[course]
if marks >= cutoff:
    print(f"Congratulations! You are eligible for {full}.")
else:
    print(f"Under score! You need at least {cutoff} marks for {short}.")
```
